## Project: Insurance Claim Fraud Detection & Investigation Prioritization

### Business scenario
SecureSure Insurance processes a large volume of motor insurance claims. The fraud investigation team has limited capacity, so the company needs a risk-ranking system to prioritize claims for manual review. The system must not automatically reject claims.

### Business objective
Develop an ML-assisted system that ranks insurance claims by fraud risk, enabling investigators to identify as many fraudulent claims as possible within the monthly investigation capacity of 200 claims, while minimizing unnecessary investigation of legitimate claims.

### Assumptions for the case study (to be validated)
- 10,000 claims per month
- Estimated fraud prevalence: 6%
- Investigation capacity: 200 claims/month
- Investigation cost: ₹2,000 per claim
- Average fraudulent claim payout: ₹50,000

These are scenario assumptions, not facts from the Kaggle dataset.

### KPI hierarchy
1. **Primary:** Precision@200 — proportion of selected claims that are fraudulent.
2. **Secondary:** Recall@200 — proportion of all fraud cases captured in the top 200.
3. **Constraint:** 200 investigations per month.
4. **Guardrails:** monitor false-positive burden, processing delays, complaints and fairness.
5. **Business outcome:** confirmed fraud value and net benefit after investigation costs.

### Modeling principle
Fraud risk scores support human investigation. A model score is not proof of fraud and must not be used as an automatic rejection decision.
